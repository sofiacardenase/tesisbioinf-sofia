```r
library(immuneSIM)

n_seqs <- 30

sim_repertoire <- immuneSIM(
  name_repertoire   = paste0("sim_rep_", n_seqs),
  number_of_seqs    = n_seqs,
  species           = "hs",
  receptor          = "ig",
  chain             = "h",
  verbose           = TRUE,
  equal_cc          = TRUE,
  shm               = "none",
  shm.prob          = 0,
  vdj_noise         = 0,
  max_cdr3_length   = 15,
  min_cdr3_length   = 15
)

# Guardar objeto simulado (opcional)
save(sim_repertoire, file = "sim_repertoire.RData")

# Mostrar gráficos en pantalla
plot_report_repertoire(sim_repertoire)   # <-- sin output_dir
```
