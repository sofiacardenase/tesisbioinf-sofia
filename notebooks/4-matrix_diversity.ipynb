# Matriz de métricas de diversidad

Esta matriz reúne los resultados de las métricas de diversidad obtenidas a partir de distintos escenarios y diferentes cantidades de secuencias simuladas.  

Cada fila representa una muestra o escenario específico, mientras que las columnas corresponden a las métricas calculadas, incluyendo riqueza, índices de diversidad, dominancia, uniformidad y estimadores de riqueza.  

La integración de todas las tablas en una sola matriz permite comparar de manera conjunta los patrones de diversidad entre escenarios y profundidades de secuenciación.

```python
import glob
import polars as pl

# Buscar todos los TSV
files = glob.glob(
    "/Users/catg/Desktop/SOFIAC/Gitsofia/tesisbioinf-sofia/results/diversity_metrics/*.tsv"
)

# Verificar que existan archivos
if not files:
    raise FileNotFoundError(
        "No se encontraron archivos TSV en diversity_metrics"
    )

# Leer y unir
df_diversity = pl.concat(
    [pl.read_csv(f, separator="\t") for f in files],
    how="vertical_relaxed"
)

# Ordenar
df_diversity = df_diversity.sort(
    by=["condition", "size"],
    descending=[False, False]
)

# Verificar orden
print(
    df_diversity.select(
        ["sample_id", "condition", "size"]
    )
)

# Guardar matriz ordenada
df_diversity.write_csv(
    "/Users/catg/Desktop/SOFIAC/Gitsofia/tesisbioinf-sofia/results/diversity_matrix/diversity_matrix.tsv",
    separator="\t"
)
```
