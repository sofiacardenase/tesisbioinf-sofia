### Introducción de SHM con SHazaM

Para generar el escenario D con hipermutación somática, se utilizó un enfoque post hoc, incorporando SHM sobre las secuencias previamente simuladas con `immuneSIM` sin mutaciones.

1. **Carga del repertorio sin SHM:** `readDNAStringSet()` lee el archivo FASTA y almacena las secuencias como un objeto `DNAStringSet`.

2. **Simulación de SHM:** `shmulateSeq()` introduce mutaciones somáticas en cada secuencia. Se utiliza una frecuencia de mutación de 12% (`numMutations = 0.12`, `frequency = TRUE`) y el modelo humano de targeting `HH_S5F`.

3. **Reproducibilidad:** `set.seed(123)` permite obtener los mismos resultados al repetir la simulación.

4. **Conversión y nombres:** `DNAStringSet()` convierte las secuencias resultantes al formato utilizado por Biostrings y se conservan los nombres originales.

5. **Generación del FASTA:** `writeXStringSet()` guarda las secuencias con SHM en un nuevo archivo FASTA.

```r
library(Biostrings)
library(immuneSIM)
library(here)
```

```r
# SIMULACIÓN ESCENARIO 4
# Cargar librerías


# Parámetro: número de secuencias

n_seqs <-100  # Cambiar a 10, 100, 1000, etc.

# Simular repertorio D folicular (centro germinal)  alta SHM,
# selección por afinidad, expansión marcada de pocos clones, CDR3 con rango amplio en R

sim_repertoire <- immuneSIM(
  name_repertoire = paste0("sim_rep_", n_seqs),
 number_of_seqs = n_seqs,
  species = "hs",
  receptor = "ig",
  chain = "h",
  verbose= TRUE,
  equal_cc = FALSE,
  user_defined_alpha = 1.5,  
  shm = "none",
  shm.prob = 0,
  vdj_noise = 0,
  max_cdr3_length =15, 
  min_cdr3_length =13,
  )

# Preparar secuencias para FASTA

dna_sequences <- Biostrings::DNAStringSet(sim_repertoire$sequence)
names(dna_sequences) <- paste0("seq", seq_along(dna_sequences))



# Guardar archivo FASTA en 'data/input'

out_file <- here("..", "data", "input",
                 paste0("repertorio_D_shm0_insilico_", n_seqs, "_seqs.fasta"))

writeXStringSet(dna_sequences, filepath = out_file)
message("Archivo guardado en: ", out_file)
```

```r
# LEER FASTA SIN SHM

input_file <- here(
  "..", "data", "input",
  "repertorio_D_shm0_insilico_100_seqs.fasta"
)

dna_sequences <- Biostrings::readDNAStringSet(input_file)


# INTRODUCIR SHM CON SHMULATESEQ

set.seed(123)

dna_sequences_shm <- sapply(
  dna_sequences,
  function(seq) {
    
    shazam::shmulateSeq(
      sequence = as.character(seq),
      numMutations = 0.12,
      frequency = TRUE,
      targetingModel = shazam::HH_S5F
    )
    
  }
)


# Convertir el resultado final a DNAStringSet

dna_sequences_shm <- Biostrings::DNAStringSet(dna_sequences_shm)


# Mantener los mismos nombres

names(dna_sequences_shm) <- names(dna_sequences)


# Guardar FASTA con SHM

out_file <- here(
  "..", "data", "input",
  "repertorio_D_insilico_100_seqs_shm12.fasta"
)

Biostrings::writeXStringSet(
  dna_sequences_shm,
  filepath = out_file
)

message("Archivo guardado en: ", out_file)
```
