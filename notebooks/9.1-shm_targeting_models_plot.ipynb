```python
import matplotlib.pyplot as plt
from matplotlib.lines import Line2D

import matplotlib.image as mpimg
```

```python
# Esta lista contiene los nombres de los archivos de imagen que se van a mostrar (en orden)
# Es decir la fila uno columna uno mostrara el grafico de la muestra: C_800_targeting_C.png
# La fila uno columna dos {basedir}/C_1600_targeting_C.png
# y asi.. tiene relacion directa con la cantidad de filas y columnas que defines en la siguiente celda

# Los nombres de las imagenes son los nombres de los archivos que estan en la carpeta (tu debes seleccionar cuales mostrar mediante el arreglo)
# tesisbioinf-sofia/results/shm_models/targeting/plots
basedir = "../results/shm_models/targeting/plots"

images = [#f"{basedir}/A_800_targeting_C.png",f"{basedir}/A_1600_targeting_C.png",f"{basedir}/A_3200_targeting_C.png",f"{basedir}/A_6400_targeting_C.png",
          #f"{basedir}/B_800_targeting_C.png",f"{basedir}/B_1600_targeting_C.png",f"{basedir}/B_3200_targeting_C.png",f"{basedir}/B_6400_targeting_C.png",
          f"{basedir}/C_800_targeting_C.png",f"{basedir}/C_1600_targeting_C.png",f"{basedir}/C_3200_targeting_C.png",f"{basedir}/C_6400_targeting_C.png",
          f"{basedir}/D_800_targeting_C.png",f"{basedir}/D_1600_targeting_C.png",f"{basedir}/D_3200_targeting_C.png",f"{basedir}/D_6400_targeting_C.png",
          f"{basedir}/E_800_targeting_C.png",f"{basedir}/E_1600_targeting_C.png",f"{basedir}/E_3200_targeting_C.png",f"{basedir}/E_6400_targeting_C.png",
                   

          ]

# Cuando descomentes los estadios A y B debes cambiar la cantidad de filas a 3 y la cantidad de columnas a 5 
```

```python
# En esta celda crearemos la figura e insertaremos las imagenes ya generadas en la carpeta de resultados.

## Las siguientes variables permiten definir la cantidad de filas y columnas que tendra la figura final.
## (tiene que tener sentido con la cantidad de elementos que hay en la lista images)
n_filas = 3
n_columnas = 4
fig, axes = plt.subplots(n_filas, n_columnas, figsize=(20, 10))

## Lo siguiente es simplemente para ponerle un titulo a cada columna, entonces lo que debes hacer es:
## - tomar los nombres de las imagenes de la primera fila y ponerles en orden
## (puedes comentarlo en caso de que no lo necesites) 
for ax, ruta in zip(axes.flat, images):
    if("C_800_targeting_C.png" in ruta):

        # titulo para la columna
        ax.text(0.5, 1.05, "800",transform=ax.transAxes,ha="center",va="top",fontsize=12,fontweight="medium")

        #titulo para la fila
        ax.text(0.15, 0.5, "Dataset C",
                transform=ax.transAxes,
                rotation=90,
                ha="center", va="center",
                fontsize=12, fontweight="medium")
    if("C_1600_targeting_C.png" in ruta):
        ax.text(0.5, 1.05, "1600",transform=ax.transAxes,ha="center",va="top",fontsize=12,fontweight="medium")
    if("C_3200_targeting_C.png" in ruta):
        ax.text(0.5, 1.05, "3200",transform=ax.transAxes,ha="center",va="top",fontsize=12,fontweight="medium")
    if("C_6400_targeting_C.png" in ruta):
        ax.text(0.5, 1.05, "6400",transform=ax.transAxes,ha="center",va="top",fontsize=12,fontweight="medium")
    if("D_800_targeting_C.png" in ruta):
        #titulo para la fila
        ax.text(0.15, 0.5, "Dataset D",
                transform=ax.transAxes,
                rotation=90,
                ha="center", va="center",
                fontsize=12, fontweight="medium")
        
    if("E_800_targeting_C.png" in ruta):
        #titulo para la fila
        ax.text(0.15, 0.5, "Dataset E",
                transform=ax.transAxes,
                rotation=90,
                ha="center", va="center",
                fontsize=12, fontweight="medium")
    img = mpimg.imread(ruta)
    ax.imshow(img)
    ax.axis("off")   # quitar ejes
    
plt.subplots_adjust(
    left=0.02,
    right=0.98,
    bottom=0.02,
    top=0.92,
    wspace=0.02,
    hspace=0.05
)

# Yo habia agregado una linea horizontal en el medio de la figura para separar las tres filas,
#  pero si no quieres que aparezca puedes comentar la siguiente parte del codigo.

# line = Line2D(
#     [0.05, 0.95],  # punto inicial y final en el eje X 
#     [0.30, 0.30],  # punto inicial y final en el eje Y
#     transform=fig.transFigure,
#     color="black",
#     linewidth=1
# )

# fig.add_artist(line)

# Guardar la figura (puedes usar diferentes formatos como .png, .jpg, .svg, etc.)
fig.savefig(
    "figura_completa_all.png",
    format = "png",  
    bbox_inches="tight",
    pad_inches=0,
    dpi=300
)
```
