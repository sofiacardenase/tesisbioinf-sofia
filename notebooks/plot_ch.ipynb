```r
library(readxl)
library(ggplot2)

# Cargar archivo Excel
datos <- read_excel("/Users/catg/Desktop/chiorazzi1.xlsx")

# Gráfico
ggplot(datos, aes(
  x = factor(`Number of cells in RNAlater`),
  y = `IgM reads`,
  fill = factor(`Number of cells in RNAlater`)
)) +
  
  # Boxplot
  geom_boxplot(
    width = 0.6,
    alpha = 0.8,
    outlier.shape = NA
  ) +
  
  # Mostrar muestras individuales
  geom_jitter(
    width = 0.08,
    size = 2,
    alpha = 0.7
  ) +
  
  # Umbral de 1.000 reads
  geom_hline(
    yintercept = 1000,
    linetype = "dashed",
    linewidth = 0.8
  ) +
  
  # Evitar notación científica en el eje X
  scale_x_discrete(
    labels = function(x) {
      format(
        as.numeric(x),
        big.mark = ",",
        scientific = FALSE
      )
    }
  ) +
  
  # Etiquetas
  labs(
    x = "Number of cells in RNAlater",
    y = "IgM reads"
  ) +
  
  # Estilo
  theme_classic() +
  theme(
    legend.position = "none",
    axis.title = element_text(size = 14),
    axis.text = element_text(size = 12),
    axis.text.x = element_text(
      angle = 45,
      hjust = 1
    )
  )
```

```r
library(readxl)
library(ggplot2)

datos1 <- read_excel("/Users/catg/Desktop/chiorazzi1.xlsx")
datos2 <- read_excel("/Users/catg/Desktop/chiorazzi2.xlsx")

datos1 <- datos1[c("Number of cells in RNAlater", "IgM reads")]
datos2 <- datos2[c("Number of cells in RNAlater", "IgM reads")]

datos1$Tanda <- "May 2025"
datos2$Tanda <- "January 2026"

datos <- rbind(datos1, datos2)

ggplot(datos, aes(
  x = `Number of cells in RNAlater`,
  y = `IgM reads`,
  color = Tanda
)) +
  geom_point(size = 3, alpha = 0.8) +
  geom_smooth(method = "lm", se = FALSE) +
  geom_hline(yintercept = 1000, linetype = "dashed", color = "black") +
  scale_x_log10(
    breaks = c(100, 1000, 5000, 10000, 50000, 100000, 500000, 1000000),
    labels = scales::comma
  ) +
  scale_y_log10(
    breaks = c(100, 1000, 5000, 10000, 50000, 100000),
    labels = scales::comma
  ) +
  labs(
    x = "Number of cells in RNAlater",
    y = "IgM reads",
    color = "Sequencing batch"
  ) +
  theme_classic()
```
