```python
escenarios <- c("A", "B", "C", "D", "E")

profundidades <- c(100, 200, 400, 800, 1600, 3200,
                   6400, 12800, 25600, 51200, 102400)

set.seed(123)

for (esc in escenarios) {
  
  datos <- read.delim(paste0(esc, "_102400.tsv"))
  
  orden <- sample(1:nrow(datos))
  
  for (n in profundidades) {
    
    datos_sub <- datos[orden[1:n], ]
    
    write.table(datos_sub,
                paste0(esc, "_", n, ".tsv"),
                sep = "\t",
                row.names = FALSE,
                quote = FALSE)
  }
}
```
