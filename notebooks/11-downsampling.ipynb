# Downsampling de los repertorios simulados

El **downsampling** se realiza a partir de los archivos `clone-pass.tsv` obtenidos después del procesamiento con **Change-O**, que contienen las secuencias asignadas a clones.

El código toma el repertorio completo de **102.400 secuencias** de cada escenario (A–E) y genera un **orden aleatorio de todas las filas** mediante `sample()`. Luego, utilizando este mismo orden aleatorio, selecciona las primeras `n` secuencias para cada profundidad de muestreo: **100, 200, 400, 800, 1.600, 3.200, 6.400, 12.800, 25.600, 51.200 y 102.400**.

Por ejemplo, para una profundidad de 100 se seleccionan las primeras 100 secuencias del orden aleatorizado; para 400, las primeras 400. De esta manera, los subconjuntos son **acumulativos** y se obtienen a partir del mismo repertorio original de 102.400 secuencias.

Cada subconjunto se guarda como un archivo `.tsv` independiente en `data/output/downsampling`, para posteriormente calcular las **métricas de diversidad** a las distintas profundidades sin volver a generar las simulaciones con `immuneSIM`.

## Funciones usadas

* `c()` → define los escenarios y las profundidades.
* `set.seed()` → permite reproducir la selección aleatoria.
* `for()` → recorre los escenarios y las profundidades.
* `read.delim()` → lee los archivos `.tsv`.
* `file.path()` → construye las rutas de entrada y salida.
* `paste0()` → construye los nombres de los archivos.
* `nrow()` → obtiene el número de filas del archivo.
* `sample()` → genera un orden aleatorio de las filas.
* `write.table()` → guarda cada subconjunto como un archivo `.tsv`.



```r
escenarios <- c("A", "B", "C", "D", "E")

profundidades <- c(100, 200, 400, 800, 1600, 3200,
                   6400, 12800, 25600, 51200, 102400)

carpeta_entrada <- "/Users/catg/Desktop/SOFIAC/Gitsofia/tesisbioinf-sofia/data/output"

carpeta_salida <- "/Users/catg/Desktop/SOFIAC/Gitsofia/tesisbioinf-sofia/data/output/downsampling"

set.seed(123)

for (esc in escenarios) {
  
  datos <- read.delim(file.path(
    carpeta_entrada,
    paste0("repertorio_", esc, "_insilico_102400_seqs_clone-pass.tsv")
  ))
  
  orden <- sample(1:nrow(datos))
  
  for (n in profundidades) {
    
    datos_sub <- datos[orden[1:n], ]
    
    write.table(
      datos_sub,
      file.path(
        carpeta_salida,
        paste0("repertorio_", esc, "_insilico_", n, "_seqs_clone-pass.tsv")
      ),
      sep = "\t",
      row.names = FALSE,
      quote = FALSE
    )
  }
}
```
