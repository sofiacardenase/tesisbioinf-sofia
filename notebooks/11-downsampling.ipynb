```python
escenarios <- c("A", "B", "C", "D", "E")

profundidades <- c(100, 200, 400, 800, 1600, 3200,
                   6400, 12800, 25600, 51200, 102400)

carpeta <- "/Users/catg/Desktop/SOFIAC/Gitsofia/tesisbioinf-sofia/data/output/downsampling"

set.seed(123)

for (esc in escenarios) {
  
  datos <- read.delim(paste0(
    "repertorio_", esc, "_insilico_102400_seqs_clone-pass.tsv"
  ))
  
  orden <- sample(1:nrow(datos))
  
  for (n in profundidades) {
    
    datos_sub <- datos[orden[1:n], ]
    
    write.table(
      datos_sub,
      file.path(
        carpeta,
        paste0("repertorio_", esc, "_insilico_", n, "_seqs_clone-pass.tsv")
      ),
      sep = "\t",
      row.names = FALSE,
      quote = FALSE
    )
  }
}
```
